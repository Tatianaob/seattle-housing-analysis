# Seattle Housing Data — Data Cleaning

This notebook prepares the Seattle housing dataset for analysis and
machine learning.

The cleaning process will include:
- Checking for missing values
- Checking for duplicate records
- Identifying invalid values
- Investigating outliers
- Correcting data types where necessary
- Creating a cleaned dataset

```python
import pandas as pd
import numpy as np
```

## 1. Load the Dataset

We will load the original raw dataset and create a separate working
DataFrame for cleaning.

```python
DATA_PATH = "../data/raw/kc_house_data.csv"


df = pd.read_csv(DATA_PATH)

df_clean = df.copy()

df_clean.info()
df_clean.dtypes
```

## 2. Missing Values

We will identify columns containing missing values before making
any changes to the dataset.

```python
missing_values = df_clean.isnull().sum()

missing_values[missing_values > 0]

missing_percentage = (
    df_clean.isnull().sum() / len(df_clean) * 100
)

missing_percentage[missing_percentage > 0]
```

## 3. Duplicate Records

```python
duplicate_count = df_clean.duplicated().sum()

print(f"Number of duplicate rows: {duplicate_count}")
```

```python
# Check for invalid values
print("Price <= 0:", (df_clean["price"] <= 0).sum())
print("Living space <= 0:", (df_clean["sqft_living"] <= 0).sum())
print("Bedrooms <= 0:", (df_clean["bedrooms"] <= 0).sum())
print("Bathrooms <= 0:", (df_clean["bathrooms"] <= 0).sum())
```

```python
# Investigate extreme values

df_clean[[
    "price",
    "bedrooms",
    "bathrooms",
    "sqft_living",
    "sqft_lot"
]].describe()

df_clean.nlargest(10, "price")[
    ["price", "bedrooms", "bathrooms", "sqft_living"]
]

df_clean.nlargest(10, "sqft_living")[
    ["price", "bedrooms", "bathrooms", "sqft_living"]
]
```
