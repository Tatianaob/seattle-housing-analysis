```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DATA_PATH = "../data/processed/kc_house_data_clean.csv"

df = pd.read_csv(DATA_PATH)

df.head()


df.shape
df.info()
```

## 1. Characteristics of the Most Expensive Homes

We will examine the characteristics of the most expensive properties
in the dataset.

```python
top_10_expensive = df.nlargest(10, "price")

top_10_expensive[
    [
        "price",
        "bedrooms",
        "bathrooms",
        "sqft_living",
        "sqft_lot",
        "grade",
        "waterfront",
        "zipcode"
    ]
]
top_10_expensive[
    [
        "price",
        "bedrooms",
        "bathrooms",
        "sqft_living",
        "sqft_lot",
        "grade"
    ]
].mean()
```
